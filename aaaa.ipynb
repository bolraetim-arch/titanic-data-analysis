```python
from sklearn.model_selection import train_test_split

# Take a fresh copy of the original dataframe
df_copy = df.copy()

# Split the dataframe into 80% training and 20% testing
train_df, test_df = train_test_split(df_copy, test_size=0.2, random_state=42)

# Print the shapes of the datasets
print(f"Original dataset shape: {df_copy.shape}")
print(f"Training dataset shape: {train_df.shape}")
print(f"Testing dataset shape:  {test_df.shape}")
```

```python
# Check missing values in the Training set
print("--- Missing Values in Training Set ---")
print(train_df.isna().sum())
print("\n" + "="*40 + "\n")

# Check missing values in the Testing set
print("--- Missing Values in Testing Set ---")
print(test_df.isna().sum())
```

```python
# Print unique values of the Cabin column in the Training set
print("Unique Cabins in Training Set:")
print(train_df['Cabin'].unique())
```

```python
# Convert Cabin to a binary feature (1 if cabin is known, 0 if missing/NaN)
train_df['Has_Cabin'] = train_df['Cabin'].notna().astype(int)
test_df['Has_Cabin'] = test_df['Cabin'].notna().astype(int)

# Drop the original Cabin column since it's no longer needed
train_df = train_df.drop(columns=['Cabin'])
test_df = test_df.drop(columns=['Cabin'])

# Verify the changes by printing the first few rows of the new column
print("--- Training Set Preview ---")
print(train_df[['Has_Cabin']].head())
print("\n--- Testing Set Preview ---")
print(test_df[['Has_Cabin']].head())
```

```python
# 1. Calculate the median age from the Training set ONLY
age_median = train_df['Age'].median()

# 2. Fill the missing values in both sets using the calculated training median
train_df['Age'] = train_df['Age'].fillna(age_median)
test_df['Age'] = test_df['Age'].fillna(age_median)

# 3. Verify that there are no more missing values in the Age column
print(f"Calculated Training Median Age: {age_median}")
print(f"Missing values in Train 'Age': {train_df['Age'].isna().sum()}")
print(f"Missing values in Test 'Age':  {test_df['Age'].isna().sum()}")
```

```python
# 1. Calculate the mode of Embarked from the Training set ONLY
embarked_mode = train_df['Embarked'].mode()[0]

# 2. Fill the missing values in both sets using the calculated training mode
train_df['Embarked'] = train_df['Embarked'].fillna(embarked_mode)
test_df['Embarked'] = test_df['Embarked'].fillna(embarked_mode)

# 3. Verify that there are no more missing values in the entire datasets
print(f"Calculated Training Mode for Embarked: '{embarked_mode}'")
print(f"Missing values in Train 'Embarked': {train_df['Embarked'].isna().sum()}")
print(f"Missing values in Test 'Embarked':  {test_df['Embarked'].isna().sum()}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# # 1. إكمال تعبئة Embarked بالطريقة الصحيحة (باستخدام القيمة النصية الأولى للمود)
# embarked_mode_val = train_df['Embarked'].mode().iloc[0]
# train_df['Embarked'] = train_df['Embarked'].fillna(embarked_mode_val)
# test_df['Embarked'] = test_df['Embarked'].fillna(embarked_mode_val)

# 2. دمج مجموعتي الترين والتست بعد المعالجة في متغير واحد لأجل الرسم فقط
df_processed = pd.concat([train_df, test_df], axis=0)

# 3. إعداد مساحة الرسم (3 رسومات بجانب بعضها)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# --- الرسم الأول: توزيع العمر (Age Distribution) ---
sns.kdeplot(df_copy['Age'], label='Original (Before)', fill=True, ax=axes[0], color='gray', alpha=0.5)
sns.kdeplot(df_processed['Age'], label='Processed (Train + Test)', fill=True, ax=axes[0], color='blue', alpha=0.5)
axes[0].set_title('Age Distribution: Before vs After')
axes[0].legend()

# --- الرسم الثاني: تحول الكابينة لمتغير ثنائي (Cabin Transformation) ---
# تجهيز البيانات الأصلية برسم منفصل للمقارنة البصرية
cabin_original = df_copy['Cabin'].notna().astype(int)
df_cabin_compare = pd.DataFrame({
    'Status': ['Missing (0)', 'Has Cabin (1)'] * 2,
    'Count': [
        (cabin_original == 0).sum(), (cabin_original == 1).sum(),
        (df_processed['Has_Cabin'] == 0).sum(), (df_processed['Has_Cabin'] == 1).sum()
    ],
    'Dataset': ['Original'] * 2 + ['Processed (Train+Test)'] * 2
})
sns.barplot(x='Status', y='Count', hue='Dataset', data=df_cabin_compare, ax=axes[1], palette=['gray', 'blue'])
axes[1].set_title('Cabin: Binary Transformation')

# --- الرسم الثالث: ميناء الانطلاق (Embarked) ---
# تجهيز بيانات Embarked للمقارنة قبل وبعد
df_embarked_compare = pd.DataFrame({
    'Port': list(df_copy['Embarked'].dropna()) + list(df_processed['Embarked']),
    'Dataset': ['Original'] * len(df_copy['Embarked'].dropna()) + ['Processed (Train+Test)'] * len(df_processed)
})
sns.countplot(x='Port', hue='Dataset', data=df_embarked_compare, ax=axes[2], palette=['gray', 'blue'])
axes[2].set_title('Embarked: Port Distribution')

plt.tight_layout()
plt.show()
```

```python
import numpy as np

for df_name, current_df in [('Training Set', train_df), ('Testing Set', test_df)]:
    print(f"========== Outliers in {df_name} ==========")
    
    for col in ['Age', 'Fare']:
        # حساب المدى الربيعي (IQR) بناءً على المجموعة الحالية فقط
        Q1 = current_df[col].quantile(0.25)
        Q3 = current_df[col].quantile(0.75)
        IQR = Q3 - Q1
        
        # تحديد الحدود الإحصائية
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # حساب عدد ونسبة القيم الشاذة
        outliers_mask = (current_df[col] < lower_bound) | (current_df[col] > upper_bound)
        outliers_count = outliers_mask.sum()
        outliers_percentage = (outliers_count / len(current_df)) * 100
        
        print(f"--- Column: '{col}' ---")
        print(f"  Total Rows: {len(current_df)}")
        print(f"  Outliers Count: {outliers_count}")
        print(f"  Outliers Percentage: {outliers_percentage:.2f}%")
        print(f"  Bounds: [{lower_bound:.2f} , {upper_bound:.2f}]")
    print("\n" + "="*40 + "\n")
```

```python
# Create a mapping dictionary for the Sex column
sex_mapping = {'male': 0, 'female': 1}

# Apply the mapping directly to both Train and Test sets
train_df['Sex'] = train_df['Sex'].map(sex_mapping)
test_df['Sex'] = test_df['Sex'].map(sex_mapping)

# Verify the transformation by printing a preview
print("--- Sex Column after Transformation (Train Set) ---")
print(train_df[['Sex']].head())
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# 1. إنشاء كائن الترميز وتفعيله على مجموعة التدريب فقط لمنع التسريب
# نستخدم drop='first' لتجنب إنشاء أعمدة زائدة تسبب تشويشاً رياضياً للنموذج
encoder = OneHotEncoder(sparse_output=False, drop='first', handle_unknown='ignore')

# 2. ملاءمة الكائن وتحويل بيانات التدريب
train_embarked_encoded = encoder.fit_transform(train_df[['Embarked']])

# 3. تحويل بيانات الاختبار بناءً على ما تعلمه الكائن من التدريب فقط (أمان كامل من التسريب)
test_embarked_encoded = encoder.transform(test_df[['Embarked']])

# 4. استخراج أسماء الأعمدة الجديدة (ستكون على شكل Embarked_Q و Embarked_S)
encoded_cols = encoder.get_feature_names_out(['Embarked'])

# 5. تحويل المصفوفات الناتجة إلى DataFrames مع الحفاظ على الأسطر الأصلية (Index)
train_embarked_df = pd.DataFrame(train_embarked_encoded, columns=encoded_cols, index=train_df.index)
test_embarked_df = pd.DataFrame(test_embarked_encoded, columns=encoded_cols, index=test_df.index)

# 6. دمج الأعمدة الرقمية الجديدة في الجداول وحذف عمود Embarked النصي القديم
train_df = pd.concat([train_df.drop(columns=['Embarked']), train_embarked_df], axis=1)
test_df = pd.concat([test_df.drop(columns=['Embarked']), test_embarked_df], axis=1)

# 7. طباعة عينة من الأعمدة الجديدة للتأكد من النتيجة
print("--- Embarked Columns after One-Hot Encoding (Train Set Preview) ---")
print(train_df[encoded_cols].head())
```

```python
# تحديد الأعمدة غير المفيدة للتدريب
columns_to_drop = ['PassengerId', 'Name', 'Ticket']

# إنشاء نسخ جديدة تماماً وحذف الأعمدة منها لتبقى الجداول القديمة كما هي
train_df_final = train_df.drop(columns=columns_to_drop).copy()
test_df_final = test_df.drop(columns=columns_to_drop).copy()

# طباعة للتأكد من نجاح العملية وبقاء الجداول القديمة سليمة
print("--- Final Training Set Columns (New Copy) ---")
print(train_df_final.columns.tolist())

print("\n--- Original Training Set Columns (Untouched) ---")
print(train_df.columns.tolist())
```

```python
# 1. فصل المتغير المستهدف (Survived) وميزات التدريب لمجموعة الـ Train
X_train = train_df_final.drop(columns=['Survived'])
y_train = train_df_final['Survived']

# 2. فصل المتغير المستهدف (Survived) وميزات الاختبار لمجموعة الـ Test
X_test = test_df_final.drop(columns=['Survived'])
y_test = test_df_final['Survived']

# 3. طباعة الأبعاد النهائية للتأكد من دقة الفصل والتطابق
print("--- Final Datasets Shapes for Machine Learning ---")
print(f"X_train (Training Features) shape: {X_train.shape}")
print(f"y_train (Training Target) shape:   {y_train.shape}")
print("-" * 50)
print(f"X_test (Testing Features) shape:   {X_test.shape}")
print(f"y_test (Testing Target) shape:     {y_test.shape}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Initialize the Random Forest model
# random_state=42 guarantees that you get the exact same results every time you run it
model = RandomForestClassifier(n_estimators=100, random_state=42)

# 2. Train (Fit) the model using the training data
model.fit(X_train, y_train)

# 3. Make predictions on the testing features
y_pred = model.predict(X_test)

# 4. Evaluate the model performance
accuracy = accuracy_score(y_test, y_pred)

print("========== Model Evaluation Results ==========")
print(f"Final Model Accuracy: {accuracy * 100:.2f}%\n")
print("Detailed Classification Report:")
print(classification_report(y_test, y_pred))
print("==============================================")
```
